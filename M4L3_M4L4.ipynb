# Example!

```python
# import library
import pandas as pd # dataframe

# from sklearn.model_selection import train_test_split # bagi data train-test
from sklearn.preprocessing import StandardScaler # standarin metrics
from sklearn.neighbors import KNeighborsClassifier # knn
from sklearn.metrics import confusion_matrix, accuracy_score # evaluasi


'''Data train'''

# membuat dataframe
df = pd.read_csv('train.csv') # training
# df_test = pd.read_csv('test.csv') # testing

# hapus kolom
df = df.drop([
    'id', 'bdate', 'langs', 'city', 'last_seen',
    'occupation_name', 'life_main', 'people_main',
    'career_start', 'career_end'
    ], axis=1)

# isi data kosong
# education_form --> yang kosong diisi data paling banyak (Full-time)
df['education_form'].fillna('Full-time', inplace=True)

# occupation type --> yang ksong diisi university
df['occupation_type'].fillna('university', inplace=True)

# ubah tipe data: object --> angka
# occupation type --> 'university' = 0, 'work' = 1
def change_occupation(occ):
    if occ == 'university':
        return 0
    else:
        return 1
df['occupation_type'] = df['occupation_type'].apply(change_occupation)

# education_form --> dummy variables
df[list(pd.get_dummies(df['education_form']).columns)] = pd.get_dummies(df['education_form'])
df = df.drop(['education_form'], axis=1)

# education_form --> dummy variables
df[list(pd.get_dummies(df['education_status']).columns)] = pd.get_dummies(df['education_status'])
df = df.drop(['education_status'], axis=1)


'''Data testing'''

# membuat dataframe
df_test = pd.read_csv('test.csv') # testing

# hapus kolom
df_test = df_test.drop([
    'bdate', 'langs', 'city', 'last_seen',
    'occupation_name', 'life_main', 'people_main',
    'career_start', 'career_end'
    ], axis=1)

# isi data kosong
# education_form --> yang kosong diisi data paling banyak (Full-time)
df_test['education_form'].fillna('Full-time', inplace=True)

# occupation type --> yang ksong diisi university
df_test['occupation_type'].fillna('university', inplace=True)

# ubah tipe data: object --> angka
# occupation type --> 'university' = 0, 'work' = 1
def change_occupation(occ):
    if occ == 'university':
        return 0
    else:
        return 1
df_test['occupation_type'] = df_test['occupation_type'].apply(change_occupation)

# education_form --> dummy variables
df_test[list(pd.get_dummies(df_test['education_form']).columns)] = pd.get_dummies(df_test['education_form'])
df_test = df_test.drop(['education_form'], axis=1)

# education_form --> dummy variables
df_test[list(pd.get_dummies(df_test['education_status']).columns)] = pd.get_dummies(df_test['education_status'])
df_test = df_test.drop(['education_status'], axis=1)


'''Bikin KNN'''

# pisahin x (parameter) dan y (target)
train_x = df.drop('result', axis=1)
train_y = df['result']

test_x = df_test.drop('id', axis=1)

# menstandarkan metriks
sc = StandardScaler()
train_x = sc.fit_transform(train_x)
test_x = sc.transform(test_x)

# bangun model knn
model = KNeighborsClassifier(n_neighbors=1) # jumlah tetangga = 3

# proses training
model.fit(train_x, train_y)

# proses testing

# prediksi
y_pred = model.predict(test_x)

# print(y_pred)

# simpan hasil prediksi jadi file csv
# isi --> 2 kolom
# id --> sesuai yang ada di test.csv
# result --> hasil prediksi (y_pred)
ID = df_test['id']

# hasil yang akan disimpan --> bikin jadi dataframe
result = pd.DataFrame({'id': ID, 'result': y_pred})

result.to_csv('results.csv', index=False)
```
